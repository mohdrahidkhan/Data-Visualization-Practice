```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('store dataset.csv')
```

```python
pd.DataFrame(df)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df['Category'].unique()
```

```python
revenue=df.groupby('Category')['Net_Sales_INR'].sum()
revenue
```

```python
categories=('Beauty','Clothing','Electronics','Home & Kitchen','Sports')

bars=plt.bar(categories,revenue)
plt.bar_label(bars,fmt='%.0f')
plt.xlabel('categories')
plt.ylabel('values')
plt.xticks(rotation=45)
plt.title('Categories-wise Net Sales')

plt.show()
```
