```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('store dataset.csv')
```

```python
pd.DataFrame(df)
```

```python
ax=sns.countplot(data=df,x='Payment_Method')
ax.bar_label(ax.containers[0])
plt.xlabel('Categories')
plt.xticks(rotation=45)
plt.ylabel('Count')
plt.title('Count of Payment Method')
plt.show()
```

```python
ax=sns.countplot(data=df,x='Sales_Channel')
ax.bar_label(ax.containers[0])
plt.xlabel('Sales Channels')
plt.show()
        
                 
```

```python
product_sales=df.groupby('Product')['Quantity'].sum().reset_index()
```

```python
pd.DataFrame(product_sales)
```

```python
ax=sns.barplot(data=product_sales,y='Product',x='Quantity')
ax.bar_label(ax.containers[0])
#plt.xticks(rotation=90)
plt.title('Total Product Sales')
plt.tight_layout()
plt.show()
```

```python
gross_product_sales=df.groupby('Product')['Gross_Sales_INR'].sum().reset_index()
```

```python
gross_product_sales
```

```python
plt.figure(figsize=(16,8))
ax=sns.barplot(data=gross_product_sales,x='Product',y='Gross_Sales_INR')
ax.bar_label(ax.containers[0],fontsize=13,padding=4)
plt.xticks(rotation=90,fontsize=14)
plt.yticks(fontsize=14)
plt.xlabel('Product',fontsize=16)
plt.ylabel('Gross_Sales_INR',fontsize=16)
plt.title('Product Based Gross_Sales_INR',fontsize=18)
plt.tight_layout()
plt.show()
```
