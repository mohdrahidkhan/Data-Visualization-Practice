```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('store dataset.csv')
```

```python
df
```

```python
# Catogories by gross sales
category_gross_sale=df.groupby('Category')['Gross_Sales_INR'].sum().sort_values().reset_index()
```

```python
category_gross_sale
```

```python
plt.figure(figsize=(6,6))
ax=sns.barplot(data=category_gross_sale,x='Category',y='Gross_Sales_INR')
ax.bar_label(ax.containers[0],fmt='{:,.0f}',padding=3)
plt.title('Category wise Gross_Sale')
plt.xlabel('Category',fontsize=14)
plt.ylabel('Sales Values',fontsize=14)
plt.show()
```

```python
plt.figure(figsize=(8,5))
ax=sns.barplot(data=category_gross_sale,x='Category',y='Gross_Sales_INR')
ax.bar_label(ax.containers[0],fmt='{:,.0f}',padding=3)
plt.title('Category wise Gross_Sale')
plt.xlabel('Category',fontsize=14)
plt.ylabel('Sales Values',fontsize=14)
plt.show()
```

```python
category_product = df.pivot_table(
    index='Category',
    columns='Product',
    values='Gross_Sales_INR',
    aggfunc='sum',
    fill_value=0
)
```

```python


plt.figure(figsize=(12, 6))
bottom = 0

for product in category_product.columns:

    plt.bar(
        x=category_product.index,
        height=category_product[product],
        bottom=bottom,
        label=product
    )

    bottom = bottom + category_product[product]

plt.title('Category Wise Gross Sales by Products')
plt.xlabel('Category')
plt.ylabel('Gross Sales (INR)')
plt.legend(title='Products', bbox_to_anchor=(1.02, 1))
plt.tight_layout()
plt.show()
```
