```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df=pd.read_csv('store dataset.csv')
```

```python
df
```

```python
ax=sns.histplot(data=df,x='Customer_Age',bins=20)
ax.bar_label(ax.containers[0])
plt.tight_layout()
plt.show()
```

```python
ax=sns.histplot(data=df,x='Profit_INR',bins=10)
ax.bar_label(ax.containers[0])
plt.show()
                
```

```python
ax=sns.histplot(data=df,x='Net_Sales_INR',bins=10)
ax.bar_label(ax.containers[0])
plt.show()
```

```python
#Question: Kitne  online and store ke customer hai jo returning and new hai?
customers=df.groupby(['Sales_Channel','Customer_Type']).size().reset_index(name='Customer_Count')
```

```python
pd.crosstab(df['Sales_Channel'],df['Customer_Type'])
```

```python
ax=sns.barplot(data=customers,x='Sales_Channel',y='Customer_Count',hue='Customer_Type')
for container in ax.containers:
    ax.bar_label(container)
plt.show()
```

```python
sns.countplot(data=customers,x='Sales_Channel',hue='Customer_Type')
plt.show()
```
